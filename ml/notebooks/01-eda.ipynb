```python
import pandas as pd

df = pd.read_csv("../data/train.csv")
df.head()
```

```python
df.info()
df.describe()
```

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(10, 4))
df.groupby("Sex")["Survived"].mean().plot.bar(ax=axes[0], title="Survival by Sex")
df.groupby("Pclass")["Survived"].mean().plot.bar(ax=axes[1], title="Survival by Pclass")
plt.show()
```
